### Streaming and Batch

## Streaming

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significally improves user experience, particularly for longer responses. Calling stream() returns an iterator that yeilds output chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
model.invoke("Wrtie me a 200 words of paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Wrtie me a 200 words of paragraph on Artificial Intelligence"):
    print(chunk.text, end="|", flush=True)
```

### Batch

Batching a collection of independent requests to model which can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
respones = model.batch([
    "Why do parrots have colorful features?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in respones:
    print(response)
```

```python
model.batch(
    ["Why do parrots have colorful features?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_currency': 5 # Limit to 5 parallel calls
    }
)
```
